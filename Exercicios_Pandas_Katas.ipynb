# Exercícios de Pandas

Use a célula abaixo para criar o DataFrame fornecido pelo ChatGPT.

```python
import pandas as pd

df = pd.DataFrame({
    "Aluno": ["Ana", "Bruno", "Carlos", "Daniel", "Eduarda",
               "Fernanda", "Gabriel", "Helena", "Igor", "Julia"],

    "Curso": ["Python", "Python", "Java", "Python", "Java",
              "C++", "Python", "C++", "Java", "Python"],

    "Idade": [20, 22, 19, 25, 21, 24, 23, 20, 22, 21],

    "Nota": [8.5, 6.0, 9.2, 7.1, 8.8, 5.5, 9.7, 6.8, 7.9, 10.0]
})
```

```python
df
```

## Exercício 1 — Explorando os dados

### Questão 1

Mostre as estatísticas da coluna `Nota`.

```python
df["Nota"].describe()
```

### Questão 2

Descubra quantos alunos existem em cada curso.

```python
df["Curso"].value_counts()
```

### Questão 3

Ordene os alunos pela maior nota.

```python
df.sort_values(by='Nota', ascending=False)
```

### Questão 4

Ordene os alunos pela menor idade.

```python
df.sort_values(by='Idade', ascending=True)
```

### Questão 5

Descubra qual curso possui mais alunos.

```python
df['Curso'].value_counts().idxmax()
```

## Exercício 2 — Filtros e seleção

### Questão 1

Mostre apenas os alunos com nota maior ou igual a 8.

```python
df.loc[df['Nota'] >= 8]
```

### Questão 2

Mostre apenas os alunos do curso de Python.

```python
df.loc[df['Curso'] == 'Python']
```

### Questão 3

Mostre apenas os alunos de Java com nota maior que 8.

```python
df.loc[(df['Curso'] == 'Java') & (df['Nota'] >= 8)]
```

### Questão 4

Mostre apenas as colunas `Aluno` e `Nota` dos alunos maiores de 21 anos.

```python
df.loc[df['Idade'] > 21, ['Aluno', 'Nota']]
```

### Questão 5

Mostre o nome do aluno que tirou nota 10.

```python
df.loc[df['Nota'] == 10, ['Aluno']]
```

## Exercício 3 — Agrupando informações

### Questão 1

Qual é a média das notas por curso?

```python
df.groupby('Curso')['Nota'].mean().round(2)
```

### Questão 2

Qual é a idade média dos alunos de cada curso?

```python
df.groupby('Curso')['Idade'].mean().round(2)
```

### Questão 3

Quantos alunos existem em cada curso usando `groupby()`?

```python
df.groupby('Curso')['Aluno'].count()
```

### Questão 4

Qual foi a maior nota de cada curso?

```python
df.groupby('Curso')['Nota'].max()
```

### Questão 5

Qual foi a menor nota de cada curso?

```python
df.groupby('Curso')['Nota'].min()
```
